```python
import cv2
import os

# === Configuration ===
#video_path = "VR_videos/4B 20250908.avi"   # Replace with your video path
video_path = "VR_videos/4B sucrose water air.avi"  
#video_path = "VR_videos/7E sucrose water air.avi" 
output_folder = "frames"                   # Main folder where all frames will be saved
num_frames = 20                            # Number of frames to extract
import cv2
import os

# === Get video name (without extension) ===
video_name = os.path.splitext(os.path.basename(video_path))[0]

# === Create subfolder based on video name ===
video_frame_folder = os.path.join(output_folder, video_name)
os.makedirs(video_frame_folder, exist_ok=True)

# === Load the video ===
video = cv2.VideoCapture(video_path)

if not video.isOpened():
    print("❌ Error: Could not open video.")
else:
    # Get total number of frames
    total_frames = int(video.get(cv2.CAP_PROP_FRAME_COUNT))

    # Calculate frame interval
    interval = max(total_frames // num_frames, 1)

    print(f"🎥 Video: {video_name}")
    print(f"Total frames in video: {total_frames}")
    print(f"Extracting {num_frames} frames, every {interval} frames...")

    saved = 0
    frame_idx = 0

    while saved < num_frames and video.isOpened():
        ret, frame = video.read()
        if not ret:
            break

        if frame_idx % interval == 0:
            # Use the actual frame number in the filename
            frame_filename = os.path.join(video_frame_folder, f"frame_{frame_idx:06d}.jpg")
            cv2.imwrite(frame_filename, frame)
            saved += 1

        frame_idx += 1

    video.release()
    print(f"✅ Done! Extracted {saved} frames into '{video_frame_folder}'.")
```
